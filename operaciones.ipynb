```python
import requests

BASE_URL = "http://127.0.0.1:8000"

# Definición de los lenguajes según la práctica
sigma = ["a", "b"]
L = ["", "a", "b"]
M = ["b", "aa"]
```

## 1. Operaciones de Conjuntos
En esta sección realizamos las operaciones de unión, intersección y diferencia entre los lenguajes $L$ y $M$.

```python
# L U M
res_union = requests.post(f"{BASE_URL}/lenguajes/union", json={"L": L, "M": M}).json()
print("L U M:", res_union["Resultado"])

# L \cap M
res_inter = requests.post(
    f"{BASE_URL}/lenguajes/interseccion", json={"L": L, "M": M}
).json()
print("L ∩ M:", res_inter["Resultado"])

# L - M
res_dif1 = requests.post(
    f"{BASE_URL}/lenguajes/diferencia", json={"L": L, "M": M}
).json()
print("L - M:", res_dif1["Resultado"])

# M - L
res_dif2 = requests.post(
    f"{BASE_URL}/lenguajes/diferencia", json={"L": M, "M": L}
).json()
print("M - L:", res_dif2["Resultado"])
```

## 2. Operaciones sobre Cadenas / Lenguajes
Realizamos la concatenación de lenguajes y la potencia de un lenguaje.

```python
# L . M
res_lm = requests.post(
    f"{BASE_URL}/lenguajes/concatenar", json={"L": L, "M": M}
).json()
print("L . M:", res_lm["Resultado"])

# M . L
res_ml = requests.post(
    f"{BASE_URL}/lenguajes/concatenar", json={"L": M, "M": L}
).json()
print("M . L:", res_ml["Resultado"])

# M^2
res_m2 = requests.post(
    f"{BASE_URL}/lenguajes/potencia", json={"L": M, "k": 2}
).json()
print("M^2:", res_m2["Resultado"])
```

## 3. Clausura de Kleene y Combinación
Obtenemos los primeros 8 elementos de $L^*$ y $M^*$, y finalmente resolvemos la expresión:
$$(L \cdot M) \cup (M^* \cap L^2)$$

```python
# L* (primeros 8)
res_l_star = requests.post(
    f"{BASE_URL}/lenguajes/clausura-kleene", json={"L": L, "k": 8}
).json()
print("L* (primeros 8):", res_l_star["Resultado"])

# M* (primeros 8)
res_m_star = requests.post(
    f"{BASE_URL}/lenguajes/clausura-kleene", json={"L": M, "k": 8}
).json()
print("M* (primeros 8):", res_m_star["Resultado"])

# Operación combinada: (L.M) U (M* ∩ L^2)
# 1. Obtener L.M
lm = set(res_lm["Resultado"])

# 2. Obtener L^2
l2 = set(
    requests.post(
        f"{BASE_URL}/lenguajes/potencia", json={"L": L, "k": 2}
    ).json()["Resultado"]
)

# 3. Obtener M*
m_star = set(res_m_star["Resultado"])

# 4. (M* ∩ L^2)
interseccion_part = m_star.intersection(l2)

# 5. Unión final
resultado_combinado = sorted(list(lm.union(interseccion_part)))
print("\nResultado de (L.M) U (M* ∩ L^2):", resultado_combinado)
```
